```python
!pip install -q transformers torch torchvision pillow requests

from transformers import pipeline, AutoImageProcessor
from PIL import Image
from io import BytesIO
from google.colab import files

MODELO = "aquib1011/maize-leaf-disease-convnext"

print(f"Preparando el modelo {MODELO}...")
procesador = AutoImageProcessor.from_pretrained("facebook/convnext-tiny-224")
clasificador = pipeline("image-classification", model=MODELO, image_processor=procesador)

print("\n👇 Sube la foto de la hoja de maíz desde tu computador 👇")
# Esto genera el botón de carga en la celda
archivo_subido = files.upload()

# Procesa la imagen directamente desde la memoria sin usar rutas
for nombre_archivo, contenido in archivo_subido.items():
    try:
        imagen = Image.open(BytesIO(contenido))
        resultados = clasificador(imagen)

        print(f"\n🌿 DIAGNÓSTICO DE LA HOJA: {nombre_archivo}")
        print("-" * 40)
        for res in resultados:
            print(f"🦠 {res['label'].upper()}: {res['score']*100:.2f}% de probabilidad")

    except Exception as e:
        print(f"Error al analizar la imagen: {e}")
```
